```python
import  selenium
from   selenium import webdriver
from   selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
```

```python
driver = webdriver.Chrome()
driver.get("https://www.noon.com/egypt-en/electronics-and-mobiles/television-and-video/televisions/?page=1")
time.sleep(5)
```

```python
tv_details = driver.find_elements(By.CSS_SELECTOR, "h2[data-qa='plp-product-box-name']")
print(tv_details[0].text)
tv_ratings_stars = driver.find_elements(By.CLASS_NAME, 'RatingPreviewStar-module-scss-module__zCpaOG__textCtr')
print(tv_ratings_stars[0].text)
tv_Ratings_numbers = driver.find_elements(By.CLASS_NAME, 'RatingPreviewStar-module-scss-module__zCpaOG__countCtr')
print(tv_Ratings_numbers[0].text)
tv_price_Before_Discount = driver.find_elements(By.CLASS_NAME, 'Price-module-scss-module__q-4KEG__preReductionRow')
print(tv_price_Before_Discount[0].text)
tv_price_After_Discount = driver.find_elements(By.CLASS_NAME, 'Price-module-scss-module__q-4KEG__amount')
print(tv_price_After_Discount[0].text)
tv_link = driver.find_elements(By.CSS_SELECTOR, "a[class*='productBoxLink']")
print(tv_link[0].get_attribute('href'))
tv_name = tv_details[0].text.split(',')[0]
print (tv_name)
tv_photo = driver.find_elements(By.CSS_SELECTOR, "img[class*='productImage']")
print(tv_photo[0].get_attribute('src'))
```

```python
list_of_tvs = []
for page in range(1, 7):
    driver.get(f"https://www.noon.com/egypt-en/electronics-and-mobiles/television-and-video/televisions/?page={page}")
    time.sleep(5)
    tv_details = driver.find_elements(By.CSS_SELECTOR, "h2[data-qa='plp-product-box-name']")
    tv_ratings_stars = driver.find_elements(By.CLASS_NAME, 'RatingPreviewStar-module-scss-module__zCpaOG__textCtr')
    tv_Ratings_numbers = driver.find_elements(By.CLASS_NAME, 'RatingPreviewStar-module-scss-module__zCpaOG__countCtr')
    tv_price_Before_Discount = driver.find_elements(By.CLASS_NAME, 'Price-module-scss-module__q-4KEG__preReductionRow')
    tv_price_After_Discount = driver.find_elements(By.CLASS_NAME, 'Price-module-scss-module__q-4KEG__amount')
    tv_link = driver.find_elements(By.CSS_SELECTOR, "a[class*='productBoxLink']")
    tv_photo = driver.find_elements(By.CSS_SELECTOR, "img[class*='productImage']")

    tvs = driver.find_elements(By.CSS_SELECTOR, "div[data-qa='plp-product-box']")
    for i in range(len(tvs)):
        try:
            tv_name = tv_details[i].text.split(',')[0] if i < len(tv_details) else None
            tv_rating_star = tv_ratings_stars[i].text if i < len(tv_ratings_stars) else None
            tv_rating_number = tv_Ratings_numbers[i].text if i < len(tv_Ratings_numbers) else None
            tv_price_after_discount = tv_price_After_Discount[i].text.split('\n')[0] if i < len(tv_price_After_Discount) else None
            tv_price_before_discount = tv_price_Before_Discount[i].text.split('\n')[0] if i < len(tv_price_Before_Discount) else tv_price_after_discount
            tv_url = tv_link[i].get_attribute('href') if i < len(tv_link) else None
            tv_image_url = tv_photo[i].get_attribute('src')
            website = "noon"
            
            list_of_tvs.append({
                'name': tv_name,
                'price_before_discount': tv_price_before_discount,
                'price_after_discount': tv_price_after_discount,
                'Rating': tv_rating_star,
                'Reviews_count': tv_rating_number,
                'Link': tv_url,
                'image_url': tv_image_url,
                'Source': website
        })
        except Exception as e:
            print(f"Page {page} - TV {i}: {e}")
            continue
```

```python
import pandas as pd
df = pd.DataFrame(list_of_tvs)
df.to_csv('tvs.csv', index=False)
df.head()
```

```python
new_list_of_tvs = list_of_tvs.copy()

options = Options()
options.add_argument("--lang=en")
options.add_experimental_option('prefs', {'intl.accept_languages': 'en,en_US'})

driver2 = webdriver.Chrome(options=options)

for page in range(1, 9):
    driver2.get(f"https://www.jumia.com.eg/televisions/?page={page}")
    time.sleep(3)
    driver2.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(3)

    tv_names = driver2.find_elements(By.CSS_SELECTOR, 'h3.name')
    tv_prices = driver2.find_elements(By.CSS_SELECTOR, 'div.prc')
    tv_old_prices = driver2.find_elements(By.CSS_SELECTOR, 'div.old')
    tv_ratings = driver2.find_elements(By.CSS_SELECTOR, 'div.stars')
    tv_reviews = driver2.find_elements(By.CSS_SELECTOR, 'div.rev')
    tv_links = driver2.find_elements(By.CSS_SELECTOR, 'a.core')
    tv_images = driver2.find_elements(By.CSS_SELECTOR, 'img.img')

    for i in range(len(tv_links)):
        try:
            name = tv_names[i].text if i < len(tv_names) else None
            price = tv_prices[i].text.replace('EGP', '').strip() if i < len(tv_prices) else None
            old_price = tv_old_prices[i].text.replace('EGP', '').strip() if i < len(tv_old_prices) else price
            
            if i < len(tv_ratings):
                rating_text = tv_ratings[i].text
                rating = rating_text.split('out')[0].strip() if 'out' in rating_text else None
            else:
                rating = None
            
            if i < len(tv_reviews):
                reviews_text = tv_reviews[i].text
                reviews = reviews_text.split('\n')[-1].strip('()') if '(' in reviews_text else None
            else:
                reviews = None
            
            url = tv_links[i].get_attribute('href')
            image = tv_images[i].get_attribute('data-src') if i < len(tv_images) else None

            new_list_of_tvs.append({
                'name': name,
                'price_after_discount': price,
                'price_before_discount': old_price,
                'Rating': rating,
                'Reviews_count': reviews,
                'Link': url,
                'image_url': image,
                'Source': 'jumia'
            })

        except Exception as e:
            print(f"Page {page} - TV {i}: {e}")
            continue

print(f"Total TVs: {len(new_list_of_tvs)}")
```

```python
import pandas as pd
df = pd.DataFrame(new_list_of_tvs)
df.to_csv('tvs.csv', index=False)
df.head()
```

```python
df['Reviews_count'] = df['Reviews_count'].apply(lambda x: int(float(str(x).replace('K', '')) * 1000) if isinstance(x, str) and 'K' in x else x)
```

```python

df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
df['Reviews_count'] = pd.to_numeric(df['Reviews_count'], errors='coerce')
```

```python
df.info()
```

```python
for col in df.columns:
    print(f"{col}        : {df[col].isnull().sum()} null values")
```

```python
df = df[~(df['Rating'].notna() & df['Reviews_count'].isna())]
```

```python
df['price_after_discount'] = df['price_after_discount'].str.replace(',', '').astype(float)
df['price_before_discount'] = df['price_before_discount'].str.replace(',', '').astype(float)
```
